```python
sales_revenue['InvoiceDate'] = pd.to_datetime(sales_revenue['InvoiceDate'])
sales_revenue['InvoiceMonth'] = (sales_revenue['InvoiceDate'].dt.to_period('M').astype(str))
```

```python
return_revenue['InvoiceDate'] = pd.to_datetime(return_revenue['InvoiceDate'])
return_revenue['InvoiceMonth'] = (return_revenue['InvoiceDate'].dt.to_period('M').astype(str))
```

```python
print(sales_revenue.columns)
```

```python
sales_monthly = sales_revenue.groupby('InvoiceMonth')['Revenue'].sum()
returns_monthly = return_revenue.groupby('InvoiceMonth')['Revenue'].sum()
```

```python
sales_revenue.head(10)
```

```python
sales_monthly = sales_monthly.sort_index()
```

```python
import matplotlib.pyplot as plt 

plt.figure(figsize=(12,6))
plt.plot(sales_monthly.index, sales_monthly.values)


plt.title('Monthly sales Revenue')
plt.xlabel('Month')
plt.ylabel('Revenue')

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()

```

```python
sales_country = (sales_revenue.groupby('Country')['Revenue'].sum())
```

```python
returns_country = (return_revenue.groupby('Country')['Revenue'].sum())
```

```python
net_revenue_country = (sales_country.add(returns_country, fill_value=0).sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.bar(
    net_revenue_country.index,
    net_revenue_country.values
)

plt.title('Net Revenue by Country')
plt.xlabel('Country')
plt.ylabel('Net Revenue')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
